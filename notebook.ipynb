```python
import torch
import math
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np

def generalized_spiral(K, device='cpu', offset=0):
    """More uniform spherical distribution than Fibonacci, with optional offset"""
    indices = torch.arange(K, dtype=torch.float32, device=device) + offset
    h = -1 + 2 * indices / (K - 1)
    theta = torch.acos(h)
    phi = torch.zeros(K, device=device)
    
    # Golden angle increment
    golden_angle = math.pi * (3 - math.sqrt(5))
    phi[1:] = (phi[1:] + golden_angle * indices[1:]) % (2*math.pi)
    
    # Convert to Cartesian
    x = torch.sin(theta) * torch.cos(phi)
    y = torch.sin(theta) * torch.sin(phi)
    z = torch.cos(theta)
    
    return torch.stack([x, y, z], dim=1)

def fibonacci_sphere(K, device='cpu', offset=0):
    """
    Generate K nearly-uniform points on a sphere using Fibonacci sphere method.
    
    Args:
        K: Number of points (categories)
        device: Target device for the tensor
        offset: Index offset for shifting points
        
    Returns:
        Tensor of shape (K, 3) with Cartesian coordinates on the unit sphere
    """
    indices = torch.arange(K, dtype=torch.float32, device=device) + offset
    phi = math.pi * (3. - math.sqrt(5.))  # Golden angle

    y = 1 - (indices / (K - 1)) * 2  # y goes from 1 to -1
    radius = torch.sqrt(1 - y ** 2)  # Radius at height y

    theta = phi * indices

    x = torch.cos(theta) * radius
    z = torch.sin(theta) * radius

    return torch.stack([x, y, z], dim=1)

def plot_sphere_categories(points_dict, title, filename):
    """
    Plot points on a sphere for multiple categories with different colors.
    
    Args:
        points_dict: Dictionary with category indices as keys and point tensors as values
        title: Title for the plot
        filename: Name of the file to save the plot
    """
    fig = plt.figure(figsize=(12, 6))
    
    # 3D Plot
    ax1 = fig.add_subplot(121, projection='3d')
    colors = plt.cm.tab10(np.linspace(0, 1, len(points_dict)))
    for (cat, points), color in zip(points_dict.items(), colors):
        ax1.scatter(points[:, 0], points[:, 1], points[:, 2], s=10, c=[color], label=f'Category {cat}')
    ax1.set_xlabel('X')
    ax1.set_ylabel('Y')
    ax1.set_zlabel('Z')
    ax1.set_title(f"{title} (3D)")
    
    # Polar Projection (2D)
    ax2 = fig.add_subplot(122, polar=True)
    for (cat, points), color in zip(points_dict.items(), colors):
        theta = torch.acos(points[:, 2])  # θ ∈ [0, π]
        phi = torch.atan2(points[:, 1], points[:, 0])  # ϕ ∈ [-π, π]
        ax2.scatter(phi, theta, s=10, c=[color], label=f'Category {cat}')
    ax2.set_title(f"{title} (Polar)")
    
    # Add legend
    ax2.legend(loc='center left', bbox_to_anchor=(1.1, 0.5))
    
    plt.tight_layout()
    plt.savefig(filename)
    plt.close()

# Generate pseudo-data: 10 categories, 100 points each, total 1000 points
K = 100  # Number of categories
points_per_category = 100
offsets = torch.linspace(0, 1000, K)  # Offsets to differentiate categories

# Generate points for Fibonacci and Generalized Spiral
fib_points_dict = {i: fibonacci_sphere(points_per_category, offset=offsets[i]) for i in range(K)}
spiral_points_dict = {i: generalized_spiral(points_per_category, offset=offsets[i]) for i in range(K)}

# Plot for both methods
plot_sphere_categories(fib_points_dict, "Fibonacci Sphere (10 Categories, 1000 Points)", "fibonacci_sphere_plot.png")
plot_sphere_categories(spiral_points_dict, "Generalized Spiral (10 Categories, 1000 Points)", "generalized_spiral_plot.png")
```

```python
import torch
import matplotlib.pyplot as plt

def compare_polar_views(K=100):
    fib_pts = fibonacci_sphere(K)
    spiral_pts = generalized_spiral(K)
    
    def to_polar(pts):
        theta = torch.acos(pts[:, 2])  # θ ∈ [0, π]
        phi = torch.atan2(pts[:, 1], pts[:, 0])  # ϕ ∈ [-π, π]
        return theta, phi
    
    plt.rcParams.update({'font.size': 12})
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 6), subplot_kw={'projection': 'polar'})
    
    # Fibonacci
    theta, phi = to_polar(fib_pts)
    ax1.scatter(phi, theta, s=10, alpha=0.7)
    ax1.set_title('Fibonacci Sphere', pad=20, fontsize=16)
    
    # Spiral
    theta, phi = to_polar(spiral_pts)
    ax2.scatter(phi, theta, s=10, alpha=0.7)
    ax2.set_title('Generalized Spiral', pad=20, fontsize=16)
    
    plt.show()

compare_polar_views(K=2000)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from scipy.interpolate import CubicSpline

def golden_spiral(K, dense=False):
    """Generate points using the golden spiral algorithm."""
    indices = np.arange(K) if not dense else np.linspace(0, K-1, 1000)  # Dense sampling for smooth curve
    h = -1 + 2 * indices / (K - 1)
    theta = np.arccos(h)
    golden_angle = np.pi * (3 - np.sqrt(5))
    phi = golden_angle * indices % (2 * np.pi)
    
    x = np.sin(theta) * np.cos(phi)
    y = np.sin(theta) * np.sin(phi)
    z = np.cos(theta)
    return x, y, z

# Generate original points (K=4) and dense spiral for smooth trajectory
K = 10
x, y, z = golden_spiral(K)
x_dense, y_dense, z_dense = golden_spiral(K, dense=True)

# Plot
fig = plt.figure(figsize=(12, 10))
ax = fig.add_subplot(111, projection='3d')

# Sphere surface (transparent)
u = np.linspace(0, 2*np.pi, 100)
v = np.linspace(0, np.pi, 50)
X_sphere = np.outer(np.cos(u), np.sin(v))
Y_sphere = np.outer(np.sin(u), np.sin(v))
Z_sphere = np.outer(np.ones_like(u), np.cos(v))
ax.plot_surface(X_sphere, Y_sphere, Z_sphere, color='lightyellow', alpha=0.2)

# Dense spiral trajectory (red curve)
ax.plot(x_dense, y_dense, z_dense, 'r-', linewidth=3, alpha=0.7, label='Spiral Path')

# Original points (colored by height)
categories = ['children', 'kids', 'adult', 'older_man']
sc = ax.scatter(x, y, z, c=z, cmap='viridis', s=200, edgecolor='k', depthshade=False)

# Annotate categories
# for i, (xi, yi, zi) in enumerate(zip(x, y, z)):
#     ax.text(xi, yi, zi, categories[i], fontsize=12, ha='center', va='bottom')

# Poles and styling
ax.scatter([0], [0], [1], color='red', s=100, label='North Pole')
ax.scatter([0], [0], [-1], color='blue', s=100, label='South Pole')
ax.set_box_aspect([1, 1, 1])
ax.view_init(elev=25, azim=45)
ax.legend()

plt.tight_layout()
plt.show()
```
